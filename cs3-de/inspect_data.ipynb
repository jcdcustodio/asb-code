```python
DATA_DIR = Path("data/raw").resolve()

INGESTION_TABLES = [
    "Application.Countries.csv",
    "Application.StateProvinces.csv",
    "Application.Cities.csv",
    "Sales.CustomerCategories.csv",
    "Sales.BuyingGroups.csv",
    "Application.People.csv",
    "Warehouse.Colors.csv",
    "Warehouse.PackageTypes.csv",
    "Warehouse.StockItems.csv",
    "Sales.Customers.csv",
    "Sales.Orders.csv",
    "Sales.OrderLines.csv",
    "Sales.Invoices.csv",
    "Sales.InvoiceLines.csv"
]
```

```python
def load_data(data_path: Path):
    if not data_path.exists():
        raise FileNotFoundError(f"Missing required source file: {data_path}")

    df = pl.read_csv(
        data_path,
        separator=";",
        null_values=["", "NULL", "N/A", "null", "None"],
        ignore_errors=True
    )

    return df
```

```python
data_path = DATA_DIR / INGESTION_TABLES[0]
df_test = pl.scan_csv(data_path, separator=";", null_values=["", "NULL", "N/A", "null", "None"], infer_schema_length=10000).collect()

print(data_path.name)
print(df_test.shape)
print(df_test.head())
pprint(df_test.null_count().to_dicts(), sort_dicts=False)
```
