```python
from pathlib import Path

import kagglehub
print(f"kagglehub=={kagglehub.__version__}")
```

```python
target_handle = "olistbr/brazilian-ecommerce"
target_path = Path("/Volumes/asb/olist_bronze/data_raw").resolve()

if not target_path.is_dir():
    data_dir = kagglehub.dataset_download(
        handle=target_handle,
        output_dir=target_path
    )
else:
    data_dir = target_path

files = list(Path(data_dir).glob(f"*.csv"))
files
```
